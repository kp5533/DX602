```python
#Problem 7
#Concept: Extracting substrings using string slicing.
#Task:
#Split the string color into two parts:
#    First half called first_half
#    Second half called last_half
#    Print both substrings
#Sample Output:
#first half: light
#last half: blue

# starter code:
color = 'darkblue'

# TO DO: WRITE YOUR CODE/SOLUTION BELOW

midpoint = len(color) // 2
first_half = color[:midpoint]
last_half = color[midpoint:]

print("first half:", first_half)
print("last half:", last_half)
```

```python
#Problem 11
#Concept: Using built-in functions on lists.
#len and sum
#Task:
#For the given list, calculate:
#    Number of elements and assign it to a variable length
#    Sum of the elements and assign it to a variable total
#    Mean (average) and assign it to a variable mean
#    Print all results in one sentence
#Sample Output:
#The list contains 9 elements, the sum is 40 and the mean is 3.7

# starter code:
values = [6,1,7,3,5,8,2,3,6,1]

# TO DO: WRITE YOUR CODE/SOLUTION BELOW

length = len(values)
total = sum(values)
mean = total / length

print("The list contains", length, "elements, the sum is", total, "and the mean is", mean)
```

```python
#Problem 12
#Concept: Assessing elements by position
#Task:
#From the given list, extract:
#    First element and assign it to a variable first
#    Middle element and assign it to a variable middle
#    Last element and assign it to a variable last
#    Print each value on its own line
#Sample Output:
#The first value is 7
#The middle value is 3
#The last value is 9

# starter code:
values = [6,1,7,3,5,8,2,3,6,1]

# TO DO: WRITE YOUR CODE/SOLUTION BELOW

first = values[0]
middle = values[len(values) // 2]
last = values[-1]

print("The first value is", first)
print("The middle value is", middle)
print("The last value is", last)
```

```python
#Problem 13
#Concept: Extracting sublists using slicing
#Task:
#From the list:
#  -  Extract first three elements and assign it to a variable first_three
#  -  Extract last three elements and assign it to a variable last_three
#  -  Combine them into a new list and assign it to a variable combined
#  -  Compute the mean of the combined list and assign it to a variable mean_combined
#  -  Print the combined list and the mean
#Notes:
#    You may use the built-in functions len and sum; you may not use any loops or library functions.
#Sample Output:
#The first 3 elements and last 3 elements combined are : [1, 4, 5, 3, 6, 7]
#And the mean of the combined list is 5.0

# starter code:
values = [6,1,7,3,5,8,2,3,6,1]

# TO DO: WRITE YOUR CODE/SOLUTION BELOW

first_three = values[:3]
last_three = values[-3:]
combined = first_three + last_three
mean_combined = sum(combined) / len(combined)

print("The first 3 elements and last 3 elements combined are :", combined)
print("And the mean of the combined list is", mean_combined)

```
